## Task 1: Feature Engineering

This section creates additional variables to support sales performance analysis. These variables will help identify revenue patterns, compare weekdays and weekends, and categorize products by price.

```python
# Calculate revenue
sales["revenue"] = sales["quantity"] * sales["unit_price"]

# Extract month and weekday
sales["month"] = sales["order_date"].dt.month_name()
sales["weekday"] = sales["order_date"].dt.day_name()

# Identify weekends
sales["is_weekend"] = sales["order_date"].dt.dayofweek >= 5

# Categorize products by unit price
sales["price_band"] = pd.cut(
    sales["unit_price"],
    bins=[0, 5000, 20000, float("inf")],
    labels=["Low", "Medium", "High"]
)

# Verify the new columns
display(sales.head())

print("New dataset shape:", sales.shape)
print("\nPrice band distribution:")
print(sales["price_band"].value_counts())
```

## Task 2: Branch Performance Analysis

This section evaluates branch performance using total revenue, order volume, units sold, and average revenue per order. The analysis will identify the strongest and weakest branches.

```python
branch_summary = sales.groupby("branch").agg(
    total_revenue=("revenue", "sum"),
    total_orders=("order_id", "count"),
    units_sold=("quantity", "sum"),
    average_order_revenue=("revenue", "mean")
).sort_values("total_revenue", ascending=False)

display(branch_summary)

print("\nTOP PERFORMING BRANCH:")
print(branch_summary.index[0])

print("\nLOWEST PERFORMING BRANCH:")
print(branch_summary.index[-1])
```

### Business Interpretation

Nairobi is the strongest branch by total revenue, generating KES 26,106,100 from 376 orders. Its performance is supported by the highest order volume and the highest number of units sold.

Eldoret records the lowest total revenue at KES 8,844,200 because it has the fewest orders and lowest units sold. However, Eldoret has the highest average revenue per order, suggesting that its customers spend more per transaction on average.

The business should investigate opportunities to increase order volume in Eldoret while maintaining its relatively high average order value. Nairobi should remain a key focus for sustaining revenue growth.

## Task 3: Product Performance Analysis

This section examines product performance by calculating total revenue, total units sold, and number of orders for each product. The objective is to identify the three highest-revenue products.

```python
product_summary = sales.groupby("product").agg(
    total_revenue=("revenue", "sum"),
    units_sold=("quantity", "sum"),
    total_orders=("order_id", "count")
).sort_values("total_revenue", ascending=False)

display(product_summary)

print("\nTOP 3 PRODUCTS BY REVENUE:")
display(product_summary.head(3))
```

### Business Interpretation

Laptops are the highest revenue-generating product, contributing KES 41,535,000 from 639 units sold. This indicates that laptops are a major contributor to the business's revenue.

Tablets rank second in revenue at KES 19,348,000 and have the highest unit sales at 691. This suggests strong customer demand for tablets.

Phones rank third in revenue at KES 8,880,000. Although headphones and chargers generate lower revenue, they still record substantial unit sales.

The business should prioritize laptop availability because of its strong revenue contribution while maintaining sufficient tablet inventory to meet customer demand.

## Task 4: Weekend versus Weekday Performance

This section compares sales performance between weekdays and weekends to determine whether customer purchasing behaviour changes depending on the day of the week.

```python
weekend_summary = sales.groupby("is_weekend").agg(
    total_revenue=("revenue", "sum"),
    total_orders=("order_id", "count"),
    units_sold=("quantity", "sum"),
    average_order_revenue=("revenue", "mean")
)

weekend_summary.index = ["Weekday" if not x else "Weekend"
                         for x in weekend_summary.index]

display(weekend_summary)
```

### Business Interpretation

Weekdays generate KES 51,856,200 in revenue from 723 orders, compared with KES 19,936,300 from 277 weekend orders.

Weekday sales contribute approximately 72.2% of total revenue, while weekends contribute 27.8%. This difference is largely associated with the higher number of weekday orders.

However, weekends have a slightly higher average order revenue of KES 71,972.20 compared with KES 71,723.65 on weekdays. This suggests that customers spend marginally more per order during weekends.

The business should maintain strong weekday operations while exploring weekend promotions to increase order volume.

## Task 5: Monthly Sales Performance

This section evaluates monthly revenue and order volume to identify the strongest and weakest months during the reporting period.

```python
monthly_summary = sales.groupby("month").agg(
    total_revenue=("revenue", "sum"),
    total_orders=("order_id", "count"),
    units_sold=("quantity", "sum")
)

# Arrange months chronologically
month_order = [
    "January", "February", "March",
    "April", "May", "June"
]

monthly_summary = monthly_summary.reindex(month_order)

display(monthly_summary)

best_month = monthly_summary["total_revenue"].idxmax()
worst_month = monthly_summary["total_revenue"].idxmin()

best_revenue = monthly_summary["total_revenue"].max()
worst_revenue = monthly_summary["total_revenue"].min()

percentage_difference = (
    (best_revenue - worst_revenue) / worst_revenue
) * 100

print("\nBEST MONTH:", best_month)
print("WORST MONTH:", worst_month)
print(f"Percentage difference: {percentage_difference:.2f}%")
```

### Business Interpretation

April was the best-performing month, generating KES 13,518,300 in revenue. June was the weakest month, recording KES 10,903,400.

April's revenue was approximately 23.98% higher than June's revenue. March recorded the highest order volume, with 186 orders, while May had the lowest, with 153 orders.

The results suggest that revenue performance does not depend solely on order volume. Product mix and quantities purchased also influence monthly revenue.

The business should investigate the factors that contributed to April's strong performance and consider strategies to improve revenue during weaker months.

## Task 6: Premium Online Sales in Nairobi and Mombasa

This section identifies premium-priced products purchased through the online channel in Nairobi and Mombasa. The filtered records will be exported for further business analysis.

```python
premium_online = sales[
    (sales["price_band"] == "High") &
    (sales["channel"] == "Online") &
    (sales["branch"].isin(["Nairobi", "Mombasa"]))
].copy()

print("FILTERED RECORDS:", len(premium_online))

print("\nREVENUE BY BRANCH:")
display(
    premium_online.groupby("branch").agg(
        total_orders=("order_id", "count"),
        total_revenue=("revenue", "sum")
    )
)

print("\nTOTAL REVENUE:")
print(f"KES {premium_online['revenue'].sum():,.2f}")

# Export filtered data
premium_online.to_csv(
    "outputs/premium_online_coast_capital.csv",
    index=False
)

print("\nExport completed successfully.")
```

```python
from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill, Alignment
from openpyxl.utils import get_column_letter
from openpyxl.worksheet.table import Table, TableStyleInfo

# Create workbook
wb = Workbook()

# Sheet 1: Raw sales data
ws_data = wb.active
ws_data.title = "Sales Data"

# Write headers
headers = list(branch_excel.columns)
ws_data.append(headers)

# Write records
for row in branch_excel.itertuples(index=False, name=None):
    ws_data.append(list(row))

# Format as Excel table
tab = Table(displayName="SalesTable", ref=ws_data.dimensions)
tab.tableStyleInfo = TableStyleInfo(
    name="TableStyleMedium2",
    showRowStripes=True
)
ws_data.add_table(tab)

# Sheet 2: Branch summary
ws_summary = wb.create_sheet("Branch Summary")

ws_summary.append([
    "Branch", "Total Revenue (KES)",
    "Total Orders", "Units Sold",
    "Region"
])

branches = ["Nairobi", "Mombasa", "Nakuru", "Kisumu", "Eldoret"]

for branch in branches:
    ws_summary.append([branch, None, None, None, None])

# Sheet 3: Region mapping
ws_region = wb.create_sheet("Region Mapping")
ws_region.append(["Branch", "Region"])

region_data = [
    ["Nairobi", "Central"],
    ["Mombasa", "Coast"],
    ["Nakuru", "Rift Valley"],
    ["Kisumu", "Western"],
    ["Eldoret", "North Rift"]
]

for row in region_data:
    ws_region.append(row)

# SUMIFS, COUNTIFS, and XLOOKUP formulas
for row in range(2, 7):
    ws_summary[f"B{row}"] = (
        f'=SUMIFS(SalesTable[revenue],SalesTable[branch],A{row})'
    )
    ws_summary[f"C{row}"] = (
        f'=COUNTIFS(SalesTable[branch],A{row})'
    )
    ws_summary[f"D{row}"] = (
        f'=SUMIFS(SalesTable[quantity],SalesTable[branch],A{row})'
    )
    ws_summary[f"E{row}"] = (
        f'=XLOOKUP(A{row},\'Region Mapping\'!A2:A6,'
        f'\'Region Mapping\'!B2:B6,"Not Found")'
    )

# Save workbook
wb.save("excel/duka_branch_summary.xlsx")

print("Excel workbook created successfully!")
```

```python
# Create pivot analysis using pandas
pivot_data = sales.pivot_table(
    index="branch",
    columns="channel",
    values="revenue",
    aggfunc="sum",
    fill_value=0,
    margins=True,
    margins_name="Grand Total"
)

display(pivot_data)

# Add pivot results to Excel workbook
from openpyxl import load_workbook
from openpyxl.styles import Font, PatternFill

file_path = "excel/duka_branch_summary.xlsx"

pivot_wb = load_workbook(file_path)

# Remove old sheet if it exists
if "Pivot Analysis" in pivot_wb.sheetnames:
    del pivot_wb["Pivot Analysis"]

pivot_ws = pivot_wb.create_sheet("Pivot Analysis")

# Write pivot table
pivot_ws.append(["Branch"] + list(pivot_data.columns))

for branch, values in pivot_data.iterrows():
    pivot_ws.append([branch] + list(values))

# Style header
for cell in pivot_ws[1]:
    cell.font = Font(bold=True)
    cell.fill = PatternFill(
        start_color="D9EAF7",
        end_color="D9EAF7",
        fill_type="solid"
    )

pivot_ws.freeze_panes = "B2"

# Save
pivot_wb.save(file_path)

print("Pivot Analysis sheet added successfully!")
```
